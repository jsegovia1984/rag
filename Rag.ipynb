```python
import os
import time
import subprocess
import textwrap
import gc

# ==========================================
# 1. REINICIO DEL SERVIDOR OLLAMA
# ==========================================
print("🧹 Limpiando procesos de sistema...")
subprocess.run(["pkill", "-f", "ollama"], check=False)
time.sleep(2)
subprocess.Popen(["ollama", "serve"])
time.sleep(5)
print("✅ Servidor Ollama iniciado.")

# ==========================================
# 2. IMPORTACIONES
# ==========================================
from langchain_community.document_loaders import DirectoryLoader, PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_chroma import Chroma
from langchain_ollama import OllamaEmbeddings, ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser

# ==========================================
# 3. ESTILOS DE TERMINAL
# ==========================================
class Colors:
    BLUE = '\033[94m'
    GREEN = '\033[92m'
    YELLOW = '\033[93m'
    RED = '\033[91m'
    BOLD = '\033[1m'
    ENDC = '\033[0m'

# ==========================================
# 4. CONFIGURACIÓN GENERAL
# ==========================================
FOLDER_PATH = "./downloaded_data"
EMBEDDING_MODEL = "nomic-embed-text"
LLM_MODEL = "llama3.2:3b"

if not os.path.exists(FOLDER_PATH) or not os.listdir(FOLDER_PATH):
    print(f"{Colors.RED}⚠️ No hay archivos en '{FOLDER_PATH}'.{Colors.ENDC}")
    exit(1)

# ==========================================
# 5. CLASIFICACIÓN DE DOCUMENTOS
# ==========================================
def detectar_tipo(path):
    p = path.lower()
    if "/ley" in p or "ley_" in p or "/leyes/" in p:
        return "ley"
    if "fallo" in p or "/fallos/" in p:
        return "fallo"
    if "dictamen" in p or "/dictamenes" in p:
        return "dictamen"
    return "otro"

# ==========================================
# 6. CREACIÓN DE BASE VECTORIAL
# ==========================================
def crear_base_de_datos():
    print(f"\n{Colors.YELLOW}📂 Procesando documentos...{Colors.ENDC}")

    loader = DirectoryLoader(
        FOLDER_PATH,
        glob="**/*.pdf",
        loader_cls=PyPDFLoader
    )

    docs = loader.load()

    # Eliminar páginas vacías (PDFs escaneados)
    docs = [d for d in docs if d.page_content.strip()]

    if not docs:
        raise ValueError("No se pudo extraer texto de los PDFs.")

    # Metadata jurídica
    for d in docs:
        d.metadata["tipo"] = detectar_tipo(d.metadata.get("source", ""))
        d.metadata["documento"] = os.path.basename(d.metadata.get("source", ""))

    splitter = RecursiveCharacterTextSplitter(
        chunk_size=4000,
        chunk_overlap=500,
        separators=["\n\n", "\n", "Art.", "ARTÍCULO", ". ", " "]
    )

    splits = splitter.split_documents(docs)
    print(f"📄 Fragmentos generados: {len(splits)}")

    embeddings = OllamaEmbeddings(model=EMBEDDING_MODEL)

    vectorstore = Chroma.from_documents(
        documents=splits,
        embedding=embeddings,
        collection_name="legal_rag_prioritario"
    )

    # Liberar memoria
    del docs, splits
    gc.collect()

    return vectorstore

# ==========================================
# 7. INICIALIZACIÓN
# ==========================================
try:
    vectorstore = crear_base_de_datos()
except Exception as e:
    print(f"{Colors.RED}Error cargando PDFs: {e}{Colors.ENDC}")
    exit(1)

# ==========================================
# 8. RETRIEVERS JERÁRQUICOS
# ==========================================
retriever_leyes = vectorstore.as_retriever(
    search_kwargs={"k": 15, "filter": {"tipo": "ley"}}
)

retriever_juris = vectorstore.as_retriever(
    search_kwargs={"k": 5, "filter": {"tipo": {"$in": ["fallo", "dictamen"]}}}
)

def format_docs(docs):
    return "\n\n".join(
        f"[Documento: {d.metadata.get('documento','desconocido')}]\n{d.page_content}"
        for d in docs
    )

def obtener_contexto(pregunta):
    leyes = retriever_leyes.invoke(pregunta)
    juris = retriever_juris.invoke(pregunta)
    return format_docs(leyes + juris)  # LEYES SIEMPRE PRIMERO

# ==========================================
# 9. PROMPT LEGAL
# ==========================================
template = """
Eres un abogado experto en Derecho Argentino.

REGLA JERÁRQUICA OBLIGATORIA:
1. Prioriza SIEMPRE la ley
2. Usa fallos y dictámenes solo para interpretar o complementar la ley.
3. Si hay contradicción, prevalece la ley.
4. Indica de qué documento surge cada afirmación.

MODO TRANSCRIPCIÓN NORMATIVA (OBLIGATORIO):
- Si la pregunta refiere a un artículo específico de una ley (ej. “artículo 8”, “art. 8°”):
  1. DEBES reproducir TODOS los incisos del artículo COMPLETOS.
  2. DEBES respetar el ORDEN ORIGINAL de los incisos (a, b, c, …).
  3. NO debes interpretar, explicar, resumir ni clasificar los incisos.
  4. NO debes cambiar los títulos ni crear categorías.
  5. Cada inciso debe comenzar exactamente con su letra o número.
  6. Si falta algún inciso en el contexto, responde únicamente:
     "El artículo no se encuentra completo en el archivo cargado".


Contexto:
{context}

Pregunta:
{question}

Respuesta técnica completa (inciso por inciso):
"""

prompt = ChatPromptTemplate.from_template(template)

llm = ChatOllama(
    model=LLM_MODEL,
    temperature=0,
    num_predict=4096,
    repeat_penalty=1.2
)

chain = (
    {
        "context": obtener_contexto,
        "question": RunnablePassthrough()
    }
    | prompt
    | llm
    | StrOutputParser()
)

# ==========================================
# 10. CHAT INTERACTIVO
# ==========================================
print(f"\n{Colors.BOLD}{Colors.BLUE}--- ASISTENTE LEGAL RAG ACTIVADO ---{Colors.ENDC}")
print("Escribí tu pregunta o 'salir'.")

while True:
    try:
        print("\n" + "-" * 70)
        pregunta = input(f"{Colors.BOLD}Tu pregunta > {Colors.ENDC}").strip()

        if pregunta.lower() in ["salir", "exit", "quit"]:
            print(f"{Colors.YELLOW}Hasta luego.{Colors.ENDC}")
            break

        print(f"{Colors.YELLOW}Analizando documentos...{Colors.ENDC}")
        respuesta = chain.invoke(pregunta)

        print(f"\n{Colors.GREEN}{Colors.BOLD}ASISTENTE LEGAL:{Colors.ENDC}")
        print(textwrap.fill(respuesta, width=100, initial_indent="  ", subsequent_indent="  "))

    except Exception as e:
        print(f"{Colors.RED}⚠️ Error: {e}{Colors.ENDC}")
        break
```
